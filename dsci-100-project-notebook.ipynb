```r
### Histogram of regional win percentage (numeric variable)
win_pct_hist <- ggplot(tidy_march_madness, aes(x = reg_wins_pct)) +
  geom_histogram(binwidth = 5) +
  labs(x = "Regional Win Percentage (%)",
       y = "Number of Teams",
       title = "Distribution of Regional Win Percentage") 

win_pct_hist
```

```r
# Calculate the proportion of teams in each level of won_a_game
# (fct_recode relabels TRUE/FALSE as Yes/No so the axis is easier to read)
num_obs <- nrow(tidy_march_madness)
won_props <- tidy_march_madness |>
  mutate(won_a_game = fct_recode(won_a_game, "Yes" = "TRUE", "No" = "FALSE")) |>
  group_by(won_a_game) |>
  summarize(count = n(),
            proportion = n() / num_obs)

# Bar plot of the proportions
won_bar <- ggplot(won_props, aes(x = won_a_game, y = proportion)) +
  geom_bar(stat = "identity") +
  labs(x = "Won at Least One Tournament Game",
       y = "Proportion of Teams",
       title = "Proportion of Teams That Won a Tournament Game")

won_bar
```

Visualization 1: This scatter plot shows seed and regional win percentage, the two predictors in my research question, with each team colored by whether it won a tournament game. It helps show whether winning and non-winning teams fall in different regions of the plot. If the colors separate into distinct areas, seed and win percentage are useful for classifying teams, which supports using them in a classifier

```r
# Scatter plot of the two predictors, colored by whether the team won a game
# (fct_recode relabels TRUE/FALSE as Yes/No for the legend)
seed_winpct_scatter <- tidy_march_madness |>
  mutate(won_a_game = fct_recode(won_a_game, "Yes" = "TRUE", "No" = "FALSE")) |>
  ggplot(aes(x = seed, y = reg_wins_pct, color = won_a_game, shape = won_a_game)) +
  geom_point(alpha = 0.5) +
  labs(x = "Tournament Seed (lower = stronger)",
       y = "Regional Win Percentage (%)",
       color = "Won a Game",
       shape = "Won a Game",
       title = "Seed and Win Percentage by Tournament Outcome") 

seed_winpct_scatter
```

Visualization 2: These histograms compare the seed distribution of teams that won at least one game with those that did not. This shows directly how seed relates to the outcome I want to predict. If winners are concentrated at low seed numbers (strong seeds) and non-winners at high ones, seed is a strong predictor of tournament success.

```r
# Histograms of seed, split by whether the team won at least one game
seed_hist <- tidy_march_madness |>
  mutate(won_a_game = fct_recode(won_a_game, "Yes" = "TRUE", "No" = "FALSE")) |>
  ggplot(aes(x = seed, fill = won_a_game)) +
  geom_histogram(binwidth = 1) +
  facet_grid(rows = vars(won_a_game)) +
  labs(x = "Tournament Seed (lower = stronger)",
       y = "Number of Teams",
       fill = "Won a Game",
       title = "Seed Distribution by Tournament Outcome") 
seed_hist
```
